```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from PIL import Image
```

```python
# Load dataset
df = pd.read_excel("3 year 21-22-23 SAC Agg.xlsx")

df.head()
df.columns
df.info()
```

```python
# List intake columns
feline_cols = [
    "Intake - Relinquished By Owner Total-Feline",
    "Intake - Stray At Large Total-Feline",
    "Intake - Transferred In Total-Feline",
    "Intake - Owner Intended Euthanasia Total-Feline",
    "Intakes - Other Intakes Total-Feline"
]

canine_cols = [
    "Intake - Relinquished By Owner Total-Canine",
    "Intake - Stray At Large Total-Canine",
    "Intake - Transferred In Total-Canine",
    "Intake - Owner Intended Euthanasia Total-Canine",
    "Intakes - Other Intakes Total-Canine"
]

# Total intake
df["Total Intake"] = df[feline_cols + canine_cols].sum(axis=1)
```

```python
vmin = df["Total Intake"].min()
vmax = df["Total Intake"].max()

years = [2021, 2022, 2023]

for year in years:
    df_year = df[df["Year"] == year]

    fig = px.choropleth(
        df_year,
        locations="State",       
        locationmode="USA-states",
        color="Total Intake",
        scope="usa",
        range_color=(vmin, vmax), 
        color_continuous_scale="Cividis",
        title=f"Total Animal Intake by State ({year})",
    )

    fig.update_traces(marker_line_color="black", marker_line_width=0.5)

    fig.update_layout(
        coloraxis_colorbar=dict(
            title="Total Intake",
            tickformat=","
        )
    )

    fig.write_image(f"Heat_maps/intake_map_{year}.png", width=1200, height=700)
```

```python
# make gif
years = [2021, 2022, 2023]
 
frames = []
for year in years:
    img = Image.open(f"Heat_maps/intake_map_{year}.png")
    frames.append(img)

frames[0].save(
    "Heat_maps/animal_intake.gif",
    save_all=True,
    append_images=frames[1:],
    duration=1500,  # 1.5 seconds per frame
    loop=0
)
```
