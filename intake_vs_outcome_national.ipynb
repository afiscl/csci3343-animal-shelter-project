```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from PIL import Image
```

```python
# Load dataset
df = pd.read_excel("3 year 21-22-23 SAC Agg.xlsx")

# df.head()
# df.columns
# df.info()
```

```python
# List intake columns
feline_cols1 = [
    "Intake - Relinquished By Owner Total-Feline",
    "Intake - Stray At Large Total-Feline",
    "Intake - Transferred In Total-Feline",
    "Intake - Owner Intended Euthanasia Total-Feline",
    "Intakes - Other Intakes Total-Feline"
]

canine_cols1 = [
    "Intake - Relinquished By Owner Total-Canine",
    "Intake - Stray At Large Total-Canine",
    "Intake - Transferred In Total-Canine",
    "Intake - Owner Intended Euthanasia Total-Canine",
    "Intakes - Other Intakes Total-Canine"
]

# Total intake
df["Total Intake"] = df[feline_cols1 + canine_cols1].sum(axis=1)
```

```python
# List live outcome columns
feline_cols2 = [
    "Live Outcome - Adoption Total-Feline",
    "Live Outcome - Returned To Owner Total-Feline",
    "Live Outcome - Returned To Field Total-Feline",
    "Live Outcome - Transferred Out Total-Feline",
    "Live outcome - Other Live Outcome Total-Feline"
]

canine_cols2 = [
    "Live Outcome - Adoption Total-Canine",
    "Live Outcome - Returned To Owner Total-Canine",
    "Live Outcome - Returned To Field Total-Canine",
    "Live Outcome - Transferred Out Total-Canine",
    "Live outcome - Other Live Outcome Total-Canine"
]

# Total intake
df["Total Live Outcome"] = df[feline_cols2 + canine_cols2].sum(axis=1)
```

```python
df["Live Outcome Rate"] = df["Total Live Outcome"] / df["Total Intake"]
df["Live Outcome Rate"] = df["Live Outcome Rate"].fillna(0)
```

```python
years = [2021, 2022, 2023]

for year in years:
    df_year = df[df["Year"] == year]

    fig = px.choropleth(
        df_year,
        locations="State",
        locationmode="USA-states",
        color="Live Outcome Rate",
        scope="usa",
        color_continuous_scale="Cividis", 
        range_color=(0, 1),  
        title=f"Live Outcome Rate by State ({year})"
    )

    fig.update_traces(marker_line_color="black", marker_line_width=0.5)

    fig.update_layout(
        coloraxis_colorbar=dict(
            title="Euthanasia Rates",
            tickformat=".0%"
        )
    )

    fig.write_image(f"Heat_maps/live_outcome_rate_{year}.png", width=1200, height=700)
```

```python
# make gif
years = [2021, 2022, 2023]
 
frames = []
for year in years:
    img = Image.open(f"Heat_maps/live_outcome_rate_{year}.png")
    frames.append(img)

frames[0].save(
    "GIFs/gif_animal_live_outcome_rate.gif",
    save_all=True,
    append_images=frames[1:],
    duration=1500,  # 1.5 seconds per frame
    loop=0
)
```

```python
# Euthanasia Rates 

euth_feline = "Other Outcome - Shelter Euthanasia Total-Feline"
euth_canine = "Other Outcome - Shelter Euthanasia Total-Canine"

df["Total Euthanasia"] = df[euth_feline] + df[euth_canine]
df["Euthanasia Rate"] = df["Total Euthanasia"] / df["Total Intake"]
```

```python
years = [2021, 2022, 2023]

for year in years:
    df_year = df[df["Year"] == year]

    fig = px.choropleth(
        df_year,
        locations="State",       
        locationmode="USA-states",
        color="Euthanasia Rate",
        scope="usa",
        range_color=(0, 1), 
        color_continuous_scale="Cividis",
        title=f"Euthanasia Rates by ({year})",
    )

    fig.update_traces(marker_line_color="black", marker_line_width=0.5)

    fig.update_layout(
        coloraxis_colorbar=dict(
            title="Euthanasia Rates",
            tickformat=".0%"
        )
    )

    fig.write_image(f"Heat_maps/euthanasia_rate_{year}.png", width=1200, height=700)
```
